### Working With JSON In Python

#### Differences Between Python dict And JSON

JSON vs python dict/list in brief:

| Feature         | JSON                                    | Python dict                                |
|-----------------|-----------------------------------------|--------------------------------------------|
| Purpose         | Data exchange                           | In-memory data structure                   |
| Data types      | Limited set of data types               | Wide range of Python data types            |
| Keys            | Strings only                            | Any immutable, hashable type               |
| Ordering        | Order of keys in objects not guaranteed | Order of keys in guaranteed in 3.7+        |
| Absence of value| null                                    | None                                       |
| Booleans        | true, false                             | True, False                                |
| Comments        | not supported                           | supported                                  |
| Trailing commas | not supported                           | supported                                  |

#### The json Module And Serialization

> * serialization: python object -> JSON
> * deserialization: JSON -> python object 

```python
person_data = {
    "name": "John Doe",
    "age": 30,
    "email": "john.doe@example.com"
}
```

```python
type(person_data)
```

```python
import json
```

```python
json.dumps(person_data)
```

```python
print(json.dumps(person_data, indent=4))
```

```python
with open('person_data.json', 'w') as json_file:
    # json.dump(person_data, json_file, indent=4)
    # You can omit indent parameter if your want to send the JSON data as a payload in compact form.
    json.dump(person_data, json_file)
```

#### Deserialization

> * deserialization: JSON -> python object 

```python
json_data = '{"name": "John Doe", "age": 30, "email": "john.doe@example.com"}'
```

```python
import json
```

```python
person = json.loads(json_data)
```

```python
print(person["name"])
```

```python
with open('person_data.json', 'r') as json_file:
    loaded_data = json.load(json_file)
    print(loaded_data)
```

#### Web Requests And APIs

> * HTTP is Hypertext Transfer Protocol; used for transmitting hypertext requests and information between servers and browsers
> * client makes request; server responds with some data

```python
# urllib.request is a built-in Python module used to open URLs and download data from the internet.
# We import it once so we can use its urlopen() function to connect to a web address.
import urllib.request
```

```python
url = "https://www.andybek.com/api/data/persons"
```

```python
# urllib.request.urlopen(url) opens a connection to the web address stored in url.
# The with statement keeps the connection open while we read from it and closes it automatically afterward.
# "response" represents the data and information sent back by the web server.
with urllib.request.urlopen(url) as response:
    # response.read() downloads the response content as raw bytes.
    # .decode("utf-8") converts those bytes into readable text.
    # UTF-8 is the common text format used for JSON data.
    # The resulting text is saved in response_data for use by the rest of the notebook.
    response_data = response.read().decode("utf-8")
```

```python
response_data
```

```python
type(response_data)
```

```python
import json
```

```python
json_data = json.loads(response_data)
```

```python
print(json_data)
```

```python
type(json_data)
```

```python
type(json_data[0])
```
