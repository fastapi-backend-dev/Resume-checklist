# Higher-Order Functions in Python

A **higher-order function** is a function that uses another function as input, returns a function, or does both.

This is possible because Python treats functions as **first-class objects**: functions can be stored in variables, passed to other functions, and returned from other functions.

```python
# A function passed as an argument
def apply_twice(func, value):
    return func(func(value))

# A function returned as a result
def create_multiplier(factor):
    def multiply(value):
        return value * factor
    return multiply
```

Think of a higher-order function as a **manager**: it does not always perform the final task itself. Instead, it gives instructions to another function or creates a new function for the task.

## Advantages

- **Code reuse:** The same logic can be applied to many different functions.
- **Cleaner code:** Related behavior can be separated into small, focused functions.
- **Flexibility:** Behavior can be selected or combined while the program is running.
- **Less repetition:** Common steps such as logging, validation, and timing do not need to be copied everywhere.
- **Easy customization:** A higher-order function can return specialized functions for different inputs.
- **Good support for decorators:** Decorators are higher-order functions that add behavior to existing functions.

## Common Use Cases

| Use case | Easy explanation | Python example |
|---|---|---|
| Decorators | Add extra behavior around a function | `@login_required` |
| Data transformation | Apply the same operation to every item | `map()`, `filter()`, list comprehensions |
| Custom sorting | Describe how items should be compared | `sorted(items, key=len)` |
| Callbacks | Run a function after an event or operation | GUI events, API handlers |
| Function factories | Create functions configured for a specific task | A multiplier or tax calculator |
| Logging and timing | Measure or record function execution | A timing decorator |
| Validation | Check inputs before the real function runs | An authentication or validation decorator |

In the examples below, functions such as `loud` and `double` are **higher-order functions** because they receive a function and return a new wrapper function.

````python
def loud(func):
    def wrapper():
        result = func().upper() + "!!"
        return result
    return wrapper
````

````python
def greet():
    return "Hello, there"
````

````python
loud_greet = loud(greet)
````

````python
loud_greet()
````
